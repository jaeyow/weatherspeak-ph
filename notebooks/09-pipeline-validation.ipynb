# Pipeline Validation: OCR → Structured Metadata → Radio Scripts → TTS

**Goal:** Trace and validate the ETL data flow for 2-3 sample bulletins

**Steps validated:**
1. **Generate FRESH structured metadata** from OCR markdown using Gemma 4 E4B (via Ollama)
2. Compare new metadata quality against archived versions
3. Validate radio scripts (generated from OCR markdown)
4. Validate TTS text (generated from radio scripts)

**Requirements:**
- Ollama running locally with `gemma4:e4b` model
- OCR markdown files in `data/gemma4_results/`

**Test bulletins:**
- `PAGASA_20-19W_Pepito_SWB#01` (Severe Weather Bulletin)
- `PAGASA_22-TC02_Basyang_TCA#01` (Tropical Cyclone Alert)
- `PAGASA_25-TC22_Verbena_TCB#24` (Tropical Cyclone Bulletin)

```python
import json
from pathlib import Path
from jsonschema import validate, ValidationError
import sys
import requests
import re

# Import the schema
sys.path.insert(0, str(Path.cwd().parent))
from modal_etl.step1_ocr import PAGASA_JSON_SCHEMA

# Paths
DATA_DIR = Path.cwd().parent / "data"
GEMMA_DIR = DATA_DIR / "gemma4_results"
RADIO_DIR = DATA_DIR / "radio_bulletins"
TTS_DIR = DATA_DIR / "tts_output"

# Ollama configuration
OLLAMA_URL = "http://localhost:11434"
GEMMA_MODEL = "gemma4:e4b"

# Test bulletins
TEST_STEMS = [
    "PAGASA_20-19W_Pepito_SWB#01",
    "PAGASA_22-TC02_Basyang_TCA#01",
    "PAGASA_25-TC22_Verbena_TCB#24",
]

print(f"✓ Loaded schema with {len(PAGASA_JSON_SCHEMA['properties'])} top-level fields")
print(f"✓ Test bulletins: {len(TEST_STEMS)}")
```

## Step 1: Check Ollama Status

```python
def check_ollama_ready():
    """Check if Ollama is running and has the model."""
    try:
        resp = requests.get(f"{OLLAMA_URL}/api/tags", timeout=5)
        resp.raise_for_status()
        models = resp.json().get("models", [])
        has_model = any(GEMMA_MODEL in m.get("name", "") for m in models)
        return True, has_model
    except Exception as e:
        return False, False

ollama_ready, has_model = check_ollama_ready()
print(f"Ollama server: {'✓ Running' if ollama_ready else '✗ Not running'}")
if ollama_ready:
    print(f"Model {GEMMA_MODEL}: {'✓ Available' if has_model else '✗ Not found'}")
    if not has_model:
        print(f"  Run: ollama pull {GEMMA_MODEL}")
else:
    print("  ⚠️  Start Ollama with: ollama serve")
```

## Step 2: Generate Fresh Structured Metadata

**Important:** We generate NEW structured output from OCR markdown to validate the current extraction process.

```python
def generate_structured_metadata(markdown: str) -> dict:
    """Generate structured metadata from OCR markdown using Ollama + Gemma 4.
    
    This calls the same extraction process used in modal_etl/step1_ocr.py
    """
    if not ollama_ready:
        print("⚠️  Ollama not available - cannot generate metadata")
        return None
    
    system_prompt = (
        "You are PAGASAParseAI, an expert at converting extracted PAGASA typhoon bulletin text into structured JSON.\n\n"
        "Extract only the fields listed in the schema. Do not include full_text or any free-form text dump.\n\n"
        "CRITICAL RULES:\n"
        "- Output ONLY the JSON object. No preamble, no markdown fences, no explanation.\n"
        "- If a field cannot be determined, use null or an empty array. Never hallucinate.\n"
        "- forecast_positions must include every position shown (24h, 48h, 72h, 96h, 120h)."
    )
    
    user_prompt = (
        f"Here is the extracted text from a PAGASA bulletin:\n\n"
        f"{markdown}\n\n"
        "Convert this into the structured JSON schema."
    )
    
    payload = {
        "model": GEMMA_MODEL,
        "prompt": user_prompt,
        "system": system_prompt,
        "stream": False,
        "format": PAGASA_JSON_SCHEMA,
    }
    
    print("  Calling Ollama (this may take 30-60 seconds)...", end="", flush=True)
    
    try:
        resp = requests.post(
            f"{OLLAMA_URL}/api/generate",
            json=payload,
            timeout=300,
        )
        resp.raise_for_status()
        raw_response = resp.json()["response"]
        print(" ✓ Done")
        return json.loads(raw_response)
    except Exception as e:
        print(f" ✗ Failed: {e}")
        return None

# Generate structured metadata for each bulletin
generated_metadata = {}

for stem in TEST_STEMS:
    print(f"\n{stem}:")
    ocr_path = GEMMA_DIR / f"{stem}_markdown.md"
    
    if not ocr_path.exists():
        print("  ✗ OCR markdown not found")
        continue
    
    markdown = ocr_path.read_text(encoding="utf-8")
    print(f"  OCR markdown: {len(markdown):,} chars")
    
    metadata = generate_structured_metadata(markdown)
    if metadata:
        generated_metadata[stem] = metadata
        print(f"  ✓ Generated structured metadata")

print(f"\n{'='*80}")
print(f"Generated metadata for {len(generated_metadata)}/{len(TEST_STEMS)} bulletins")
```

### Deep Dive: Field-Level Comparison

Compare specific fields across all three bulletins to identify patterns.

```python
def get_structure_signature(obj, path=""):
    """Recursively get the structure signature of a JSON object (keys + types)."""
    if isinstance(obj, dict):
        sig = {}
        for key, value in obj.items():
            sig[key] = get_structure_signature(value, f"{path}.{key}")
        return sig
    elif isinstance(obj, list):
        if len(obj) > 0:
            # Use first item as representative
            return [get_structure_signature(obj[0], f"{path}[0]")]
        else:
            return []
    elif obj is None:
        return "null"
    else:
        return type(obj).__name__

def compare_structures(struct1, struct2, path=""):
    """Compare two structure signatures and return differences."""
    diffs = []
    
    if type(struct1) != type(struct2):
        diffs.append(f"{path}: type mismatch ({type(struct1).__name__} vs {type(struct2).__name__})")
        return diffs
    
    if isinstance(struct1, dict):
        keys1 = set(struct1.keys())
        keys2 = set(struct2.keys())
        
        missing_in_2 = keys1 - keys2
        missing_in_1 = keys2 - keys1
        
        if missing_in_2:
            diffs.append(f"{path}: keys in first but not second: {missing_in_2}")
        if missing_in_1:
            diffs.append(f"{path}: keys in second but not first: {missing_in_1}")
        
        # Recursively compare common keys
        for key in keys1 & keys2:
            new_path = f"{path}.{key}" if path else key
            diffs.extend(compare_structures(struct1[key], struct2[key], new_path))
    
    elif isinstance(struct1, list) and len(struct1) > 0 and len(struct2) > 0:
        diffs.extend(compare_structures(struct1[0], struct2[0], f"{path}[0]"))
    
    return diffs

# Get structure signatures for all three bulletins
print("="*80)
print("STRUCTURE CONSISTENCY ANALYSIS")
print("="*80)

structures = {}
for stem, metadata in generated_metadata.items():
    structures[stem] = get_structure_signature(metadata)
    print(f"\n{stem}:")
    print(f"  Top-level keys: {list(metadata.keys())}")
    print(f"  Forecast positions: {len(metadata.get('forecast_positions', []))}")
    print(f"  Affected areas signals: {list(metadata.get('affected_areas', {}).keys())}")

# Compare all pairs
print(f"\n{'='*80}")
print("PAIRWISE CONSISTENCY CHECK")
print(f"{'='*80}")

stems = list(generated_metadata.keys())
all_consistent = True

for i in range(len(stems)):
    for j in range(i + 1, len(stems)):
        stem1, stem2 = stems[i], stems[j]
        print(f"\n{stem1}")
        print(f"  vs")
        print(f"{stem2}")
        print("-" * 80)
        
        diffs = compare_structures(structures[stem1], structures[stem2])
        
        if not diffs:
            print("✅ CONSISTENT: Identical structure")
        else:
            print(f"⚠️  INCONSISTENT: {len(diffs)} difference(s)")
            for diff in diffs[:10]:  # Show first 10 diffs
                print(f"  • {diff}")
            if len(diffs) > 10:
                print(f"  ... and {len(diffs) - 10} more differences")
            all_consistent = False

print(f"\n{'='*80}")
if all_consistent:
    print("✅ ALL THREE BULLETINS HAVE CONSISTENT STRUCTURE")
else:
    print("⚠️  BULLETINS HAVE STRUCTURAL INCONSISTENCIES")
print(f"{'='*80}")
```

```python
def load_bulletin_files(stem: str) -> dict:
    """Load all available files for a bulletin."""
    files = {
        "stem": stem,
        "ocr_md": None,
        "metadata_json": None,
        "radio_en": None,
        "radio_tl": None,
        "radio_ceb": None,
        "tts_en": None,
        "tts_tl": None,
        "tts_ceb": None,
    }
    
    # OCR markdown
    ocr_path = GEMMA_DIR / f"{stem}_markdown.md"
    if ocr_path.exists():
        files["ocr_md"] = ocr_path.read_text(encoding="utf-8")
    
    # Use freshly generated metadata (from previous step)
    if stem in generated_metadata:
        files["metadata_json"] = generated_metadata[stem]
    
    # Radio scripts - look in RADIO_OUTPUT_DIR (where this notebook saves them)
    for lang in ["en", "tl", "ceb"]:
        radio_path = RADIO_OUTPUT_DIR / f"{stem}_radio_{lang}.md"
        if radio_path.exists():
            files[f"radio_{lang}"] = radio_path.read_text(encoding="utf-8")
    
    # TTS text (plain) - look in OUTPUT_DIR/tts_texts (where this notebook saves them)
    tts_dir = OUTPUT_DIR / "tts_texts"
    for lang in ["en", "tl", "ceb"]:
        # This notebook saves as: {safe_stem}_tts_{lang}.txt (with # removed)
        safe_stem = stem.replace("#", "")
        tts_path = tts_dir / f"{safe_stem}_tts_{lang}.txt"
        if tts_path.exists():
            files[f"tts_{lang}"] = tts_path.read_text(encoding="utf-8")
    
    return files

# Load all bulletins
bulletins = {stem: load_bulletin_files(stem) for stem in TEST_STEMS}

# Summary
for stem, files in bulletins.items():
    available = [k for k, v in files.items() if v is not None and k != "stem"]
    print(f"\n{stem}:")
    print(f"  Files available: {', '.join(available)}")
    print(f"  Total: {len(available)}/8 files")
```

```python
# Sentence splitting helpers from notebook 08
import time
import numpy as np
import torch
import os
from pydub import AudioSegment
from transformers import VitsModel, AutoTokenizer

def prepare_mms_sentences(text: str) -> list[tuple[str, bool]]:
    """Prepare sentences for MMS TTS (Cebuano/Tagalog).
    Returns list of (sentence, is_paragraph_end) tuples."""
    result = []
    paragraphs = [p.strip() for p in text.split("\n\n") if p.strip()]
    for paragraph in paragraphs:
        sentences = re.split(r'(?<=[.!?])\s+', paragraph)
        sentences = [s.strip() for s in sentences if s.strip()]
        for sent_idx, sentence in enumerate(sentences):
            is_last_in_para = (sent_idx == len(sentences) - 1)
            sentence = re.sub(r"\s+", " ", sentence).strip()
            if sentence:
                result.append((sentence, is_last_in_para))
    return result

def prepare_english_sentences(text: str) -> list[tuple[str, bool]]:
    """Prepare sentences for XTTS English.
    Returns list of (sentence, is_paragraph_end) tuples."""
    result = []
    paragraphs = [p.strip() for p in text.split("\n\n") if p.strip()]
    for paragraph in paragraphs:
        sentences = re.split(r'(?<=[.!?])\s+', paragraph)
        sentences = [s.strip() for s in sentences if s.strip()]
        for sent_idx, sentence in enumerate(sentences):
            is_last_in_para = (sent_idx == len(sentences) - 1)
            sentence = re.sub(r"\s+", " ", sentence).strip()
            if sentence:
                result.append((sentence, is_last_in_para))
    return result

# Prepare sentences for all TTS texts
print("Preparing sentences for synthesis...")
sentence_lists = {}

for stem, languages in generated_tts.items():
    sentence_lists[stem] = {}
    print(f"\n{stem}:")
    for lang, tts_text in languages.items():
        if lang == "en":
            sentences = prepare_english_sentences(tts_text)
        else:
            sentences = prepare_mms_sentences(tts_text)
        sentence_lists[stem][lang] = sentences
        print(f"  {lang.upper()}: {len(sentences)} sentences")

print(f"\n{'='*80}")
print("Sentence preparation complete")
print(f"{'='*80}")
```

```python
# Synthesis functions from notebook 08

def speed_change(segment: AudioSegment, speed: float) -> AudioSegment:
    """Speed up audio by resampling."""
    new_rate = int(segment.frame_rate * speed)
    return segment._spawn(segment.raw_data, overrides={"frame_rate": new_rate}).set_frame_rate(segment.frame_rate)


def synthesize_with_mms(
    sentences: list[tuple[str, bool]],
    model,
    tokenizer,
    output_path: Path,
    sentence_pause_ms: int = 150,
    paragraph_pause_ms: int = 150,
    speech_speed: float = 1.0,
) -> Path:
    """Synthesize MMS VITS sentence list to MP3 (Cebuano / Tagalog)."""
    if not sentences:
        raise ValueError("sentences list is empty")
    rate = model.config.sampling_rate
    combined = AudioSegment.empty()
    for sentence, is_paragraph_end in sentences:
        if not sentence.strip():
            continue
        inputs = tokenizer(sentence, return_tensors="pt")
        with torch.no_grad():
            waveform = model(**inputs).waveform
        pcm = (waveform.squeeze().numpy() * 32_767).clip(-32_768, 32_767).astype(np.int16)
        segment = AudioSegment(pcm.tobytes(), frame_rate=rate, sample_width=2, channels=1)
        if speech_speed != 1.0:
            segment = speed_change(segment, speech_speed)
        combined += segment
        pause_ms = paragraph_pause_ms if is_paragraph_end else sentence_pause_ms
        combined += AudioSegment.silent(duration=pause_ms, frame_rate=rate)
    combined.export(str(output_path), format="mp3", bitrate="128k")
    return output_path


def synthesize_with_xtts(
    sentences: list[tuple[str, bool]],
    tts_engine,
    output_path: Path,
    speaker: str = XTTS_SPEAKER,
    sentence_pause_ms: int = 300,
    paragraph_pause_ms: int = 500,
) -> Path:
    """Synthesize Coqui XTTS v2 sentence list to MP3 (English)."""
    if not sentences:
        raise ValueError("sentences list is empty")
    combined = AudioSegment.empty()
    for sentence, is_paragraph_end in sentences:
        if not sentence.strip():
            continue
        wav = tts_engine.tts(text=sentence, speaker=speaker, language="en")
        pcm = (np.array(wav) * 32_767).clip(-32_768, 32_767).astype(np.int16)
        segment = AudioSegment(pcm.tobytes(), frame_rate=XTTS_SAMPLE_RATE, sample_width=2, channels=1)
        combined += segment
        pause_ms = paragraph_pause_ms if is_paragraph_end else sentence_pause_ms
        combined += AudioSegment.silent(duration=pause_ms, frame_rate=XTTS_SAMPLE_RATE)
    combined.export(str(output_path), format="mp3", bitrate="128k")
    return output_path

print("✓ Synthesis functions defined")
```

```python
# Synthesize MP3s for all bulletins and languages
synthesis_config = {
    "ceb": {"sentence_pause_ms": 150, "paragraph_pause_ms": 150, "speech_speed": 1.1},
    "tl": {"sentence_pause_ms": 150, "paragraph_pause_ms": 150, "speech_speed": 1.1},
    "en": {"sentence_pause_ms": 300, "paragraph_pause_ms": 500},
}

# Output directory for MP3 files
audio_output_dir = OUTPUT_DIR / "audio"
audio_output_dir.mkdir(parents=True, exist_ok=True)

print("="*80)
print("SYNTHESIZING MP3 AUDIO FILES")
print("="*80)

synthesis_results = {}

for stem in TEST_STEMS:
    print(f"\n{stem}:")
    synthesis_results[stem] = {}
    
    for lang in ["en", "tl", "ceb"]:
        if stem not in sentence_lists or lang not in sentence_lists[stem]:
            print(f"  ✗ {lang.upper()}: No sentences to synthesize")
            continue
        
        sentences = sentence_lists[stem][lang]
        config = synthesis_config[lang]
        
        # Sanitize stem for filename
        safe_stem = stem.replace("#", "")
        output_path = audio_output_dir / f"{safe_stem}_audio_{lang}.mp3"
        
        lang_label = {"en": "English", "tl": "Tagalog", "ceb": "Cebuano"}[lang]
        print(f"  {lang.upper()}: Synthesizing {len(sentences)} sentences...", end="", flush=True)
        
        t_start = time.time()
        
        try:
            if lang == "en":
                synthesize_with_xtts(
                    sentences,
                    tts_models["en"],
                    output_path,
                    sentence_pause_ms=config["sentence_pause_ms"],
                    paragraph_pause_ms=config["paragraph_pause_ms"],
                )
            else:
                synthesize_with_mms(
                    sentences,
                    tts_models[lang],
                    tts_tokenizers[lang],
                    output_path,
                    sentence_pause_ms=config["sentence_pause_ms"],
                    paragraph_pause_ms=config["paragraph_pause_ms"],
                    speech_speed=config["speech_speed"],
                )
            
            elapsed = time.time() - t_start
            size_kb = output_path.stat().st_size // 1024
            duration_s = (size_kb * 1024 * 8) / 128_000
            
            print(f" ✓ {elapsed:.1f}s → {size_kb} KB (~{duration_s:.0f}s audio)")
            
            synthesis_results[stem][lang] = {
                "path": output_path,
                "elapsed_s": round(elapsed, 1),
                "size_kb": size_kb,
                "duration_s": round(duration_s),
            }
        except Exception as e:
            print(f" ✗ Failed: {e}")

print(f"\n{'='*80}")
total_synthesized = sum(len(langs) for langs in synthesis_results.values())
print(f"✓ Synthesized {total_synthesized} MP3 files")
print(f"  Saved to: {audio_output_dir}")
print(f"{'='*80}")
```

## Step 4: Validate Structured Metadata Against Schema

```python
def validate_metadata(metadata: dict, stem: str) -> dict:
    """Validate metadata and return quality report."""
    report = {
        "stem": stem,
        "schema_valid": False,
        "issues": [],
        "warnings": [],
    }
    
    # Schema validation
    try:
        validate(instance=metadata, schema=PAGASA_JSON_SCHEMA)
        report["schema_valid"] = True
    except ValidationError as e:
        report["issues"].append(f"Schema validation failed: {e.message}")
        return report
    
    # Data quality checks
    
    # Check coordinates
    lat = metadata.get("current_position", {}).get("latitude")
    lon = metadata.get("current_position", {}).get("longitude")
    
    if lat is not None and (lat < -90 or lat > 90):
        report["issues"].append(f"Latitude out of range: {lat}")
    
    if lon is not None and (lon < -180 or lon > 180):
        report["issues"].append(f"Longitude out of range: {lon}")
    
    # Check forecast positions
    for i, pos in enumerate(metadata.get("forecast_positions", [])):
        pos_lat = pos.get("latitude")
        pos_lon = pos.get("longitude")
        
        if pos_lat is not None and (pos_lat < -90 or pos_lat > 90):
            report["issues"].append(f"Forecast position {i} latitude out of range: {pos_lat}")
        
        if pos_lon is not None and (pos_lon < -180 or pos_lon > 180):
            report["issues"].append(f"Forecast position {i} longitude out of range: {pos_lon}")
        
        # Check for garbage labels
        label = pos.get("label", "")
        if "data_missing" in label or len(label) < 5:
            report["warnings"].append(f"Forecast position {i} has suspicious label: {label}")
        
        # Check for mangled text (excessive whitespace)
        reference = pos.get("reference", "")
        if "     " in reference or "\t" in reference:
            report["warnings"].append(f"Forecast position {i} has mangled reference text")
    
    # Check confidence
    confidence = metadata.get("confidence", 1.0)
    if confidence < 0.5:
        report["warnings"].append(f"Low confidence score: {confidence}")
    
    # Check for missing storm data
    if not metadata.get("storm", {}).get("name"):
        report["issues"].append("Missing storm name")
    
    return report

# Validate all bulletins
validation_reports = {}
for stem, files in bulletins.items():
    if files["metadata_json"]:
        validation_reports[stem] = validate_metadata(files["metadata_json"], stem)

# Display results
for stem, report in validation_reports.items():
    print(f"\n{'='*80}")
    print(f"{stem}")
    print(f"{'='*80}")
    print(f"Schema Valid: {'✓' if report['schema_valid'] else '✗'}")
    
    if report["issues"]:
        print(f"\n❌ ISSUES ({len(report['issues'])}):")
        for issue in report["issues"]:
            print(f"  • {issue}")
    
    if report["warnings"]:
        print(f"\n⚠️  WARNINGS ({len(report['warnings'])}):")
        for warning in report["warnings"]:
            print(f"  • {warning}")
    
    if not report["issues"] and not report["warnings"]:
        print("\n✅ No issues or warnings found")
```
